FINAL CHALLENGE

```python

# Step 1: Clean missing values
cleaned_df = bookings_df.fillna({
    "discount_pct": 0,
    "payment_method": "Unknown"
})

# Step 2: Remove duplicate booking IDs
deduplicated_df = cleaned_df.dropDuplicates(
    ["booking_id"]
)

# Step 3: Calculate gross and net revenue
revenue_df = deduplicated_df.withColumn(
    "gross_amount",
    F.col("tickets") * F.col("price_per_ticket")
).withColumn(
    "net_amount",
    F.col("gross_amount") *
    (1 - F.col("discount_pct") / 100)
)

# Step 4: Keep Confirmed bookings
confirmed_df = revenue_df.filter(
    F.col("booking_status") == "Confirmed"
)

# Step 5: Join cinema master
joined_df = confirmed_df.join(
    cinemas_df,
    on="cinema_id",
    how="left"
)

# Step 6: Aggregate by city and movie
performance_df = joined_df.groupBy(
    F.col("location.city").alias("city"),
    "movie_title"
).agg(
    F.sum("tickets").alias("total_tickets"),
    F.sum("net_amount").alias("net_revenue"),
    F.count("booking_id").alias("total_bookings")
)

# Step 7: Display report
display(
    performance_df.orderBy(
        F.col("net_revenue").desc()
    )
)

# Step 8: Save report as Delta table
performance_df.write.format("delta").mode(
    "overwrite"
).saveAsTable("cinema_performance_report")
```
