```python
%sql
DESCRIBE HISTORY fitness_memberships_demo;
```

```python
%sql
SELECT *
FROM fitness_memberships_demo
VERSION AS OF 1;
```

```python
%sql
SELECT
    'Current Version' AS version_type,
    COUNT(*) AS total_records
FROM fitness_memberships_demo

UNION ALL

SELECT
    'Earlier Version' AS version_type,
    COUNT(*) AS total_records
FROM fitness_memberships_demo
VERSION AS OF 1;
```
