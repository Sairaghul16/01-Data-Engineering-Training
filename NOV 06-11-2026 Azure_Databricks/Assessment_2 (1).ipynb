```python
%sql
MERGE INTO event_registrations_work AS target
USING incoming_registrations AS source
ON target.registration_id = source.registration_id

WHEN MATCHED THEN
UPDATE SET
    target.event_id = source.event_id,
    target.attendee_name = source.attendee_name,
    target.registration_date = source.registration_date,
    target.ticket_type = source.ticket_type,
    target.tickets = source.tickets,
    target.unit_price = source.unit_price,
    target.status = source.status,
    target.channel = source.channel

WHEN NOT MATCHED THEN
INSERT (
    registration_id,
    event_id,
    attendee_name,
    registration_date,
    ticket_type,
    tickets,
    unit_price,
    status,
    channel
)
VALUES (
    source.registration_id,
    source.event_id,
    source.attendee_name,
    source.registration_date,
    source.ticket_type,
    source.tickets,
    source.unit_price,
    source.status,
    source.channel
);
```

1. Permanent View
Definition: A permanent view is like a saved report. We can access it anytime, even after closing and reopening Databricks.

Real-life example: A college student report that the principal and staff can view every day.
2. Temporary View
Definition: A temporary view is like a temporary report that we can use only in the current session. Once the session ends, it disappears.

Real-life example: A teacher creates today's attendance report. After closing the session, the temporary report disappears.
3. Global Temporary View
Definition: A global temporary view is a temporary report that can be shared with multiple sessions in the same Spark application.

Real-life example: A college attendance report that multiple teachers can access while the same Spark application is running. Once the application ends, the report disappears.

- Improves data skipping.
- Helps queries that frequently filter using the ZORDER column.

```python
from pyspark.sql.functions import col, concat, lit

transaction_df = rate_df.withColumn(
    "transaction_id",
    concat(lit("TXN"), col("value").cast("string"))
).withColumn(
    "transaction_amount",
    col("value") * 100
)

display(
    transaction_df,
    checkpointLocation="/Volumes/sairaghul_training_databricks/default/data/checkpoints/rate_stream_q48"
)
```

```python
from pyspark.sql.functions import col

filtered_df = transaction_df.filter(
    col("transaction_amount") > 500
)

display(
    filtered_df,
    checkpointLocation="/Volumes/sairaghul_training_databricks/default/data/checkpoints/rate_stream_q49"
)
```

50

```python
from pyspark.sql.functions import when, col

category_df = filtered_df.withColumn(
    "transaction_category",
    when(col("transaction_amount") >= 1000, "High")
    .otherwise("Regular")
)
```

```python
from pyspark.sql.functions import count

aggregated_df = category_df.groupBy(
    "transaction_category"
).agg(
    count("*").alias("total_transactions")
)
```

```python
query = aggregated_df.writeStream \
    .format("console") \
    .outputMode("complete") \
    .option(
        "checkpointLocation",
        "/Volumes/sairaghul_training_databricks/default/data/checkpoints/q50_stream"
    ) \
    .trigger(processingTime="10 seconds") \
    .start()
```
